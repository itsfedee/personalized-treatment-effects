# 03. Cleaned data check

In this short notebook we load the dataset, apply the clean() function and briefly inspect the result.

```python
from pathlib import Path

import pandas as pd

CLEAN_PATH = Path("../data/processed/ist_clean.parquet")

df = pd.read_parquet(CLEAN_PATH)
df.shape
```

```python
df.dtypes
```

```python
# RXHEP: H recoded as M
assert (df["RXHEP"] == "M").sum() == 4856
assert (df["RXHEP"] == "H").sum() == 0

# OCCODE: 0, 8, 9 set to missing
assert df["OCCODE"].isna().sum() == 150

# pilot phase: 984 patients
assert df["pilot_phase"].sum() == 984

# RVISINF: the impossible combination set to missing
assert df["RVISINF"].isna().sum() == 1

# primary outcome: OCCODE 1-2 -> 1, 3-4 -> 0
assert (df["death_or_dependency_6m"] == 1).sum() == 12125
assert (df["death_or_dependency_6m"] == 0).sum() == 7160
assert df["death_or_dependency_6m"].isna().sum() == 150

# FDEAD: U set to missing (99 already missing + 57 U)
assert df["FDEAD"].isna().sum() == 156
```
