# Benefit and harm: average effects on every outcome

The primary outcome nets out benefits and harms. This notebook shows them separately: the average effect of each treatment on the events recorded within 14 days and on death at 6 months, in events per 1000 patients. These are plain trial averages over all patients, with no model.

```python
from pathlib import Path

import pandas as pd

from ml_med.modeling import average_effects

df = pd.read_parquet(Path("../data/processed/ist_features.parquet"))

for treatment in ("RXASP", "RXHEP"):
    print(treatment)
    display(average_effects(df, treatment).round(1))
```

## Conclusion

1. Heparin trades one harm for another. Per 1000 patients it prevents about 5 recurrent ischaemic strokes and 3 pulmonary embolisms, and causes about 8 haemorrhagic strokes and 9 major non-cerebral bleeds. Its net effect on death or dependency at 6 months is zero (+0.3, standard error 7.0).

2. Aspirin has a better balance. Per 1000 patients it prevents about 9 recurrent ischaemic strokes and 2 pulmonary embolisms, against about 5 more major bleeds and no clear increase in haemorrhagic strokes (+1.3, standard error 1.3). The net effect is about 12 fewer patients dead or dependent (standard error 7.0).

3. These are average effects. Notebooks 06 to 08 found no evidence that they differ from patient to patient.
